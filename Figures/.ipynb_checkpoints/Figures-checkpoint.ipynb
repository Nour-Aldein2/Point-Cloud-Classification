```python
def visualize_palettes(palettes):
    fig, axes = plt.subplots(len(palettes), 1, figsize=(12, 2 * len(palettes)), squeeze=False)
    axes = axes[:, 0]

    for ax, (name, colors) in zip(axes, palettes.items()):
        for i, color in enumerate(colors):
            ax.add_patch(Rectangle((i, 0), 1, 1, color=color))
            ax.text(i + 0.5, -0.12, color, ha="center", va="top", fontsize=8)

        ax.set_xlim(0, len(colors))
        ax.set_ylim(-0.5, 1)
        ax.set_title(name, loc="left", fontsize=13, fontweight="bold")
        ax.axis("off")

    plt.tight_layout()
    plt.show()

visualize_palettes(PALETTES)
```

# Statistics

```python
palette = PALETTES["house_scape"]
```

```python
class_counts = defaultdict(dict)

for d in DATA_PATH.iterdir():
    if not d.is_dir():
        continue

    class_counts[d.name]["train"] = len(list((d / "train").glob("*.off")))
    class_counts[d.name]["test"] = len(list((d / "test").glob("*.off")))
```

```python
train_color = palette[0]
test_color = palette[1]
fontsize = 10

classes = list(class_counts.keys())
labels = [c.replace("_", " ").title() for c in classes]
train_counts = [class_counts[c]["train"] for c in classes]
test_counts = [class_counts[c]["test"] for c in classes]

totals = [train + test for train, test in zip(train_counts, test_counts)]

bar_height = 0.55
group_spacing = 1.25
y = np.arange(len(classes)) * group_spacing

fig, ax = plt.subplots(figsize=(6, 8))

train_bars = ax.barh(y + bar_height / 2, train_counts, height=bar_height, color=train_color, label="Training")
test_bars = ax.barh(y - bar_height / 2, test_counts, height=bar_height, color=test_color, label="Test")

for bars, counts in [(train_bars, train_counts), (test_bars, test_counts)]:
    for bar, count, total in zip(bars, counts, totals):
        y_pos = bar.get_y() + bar.get_height() / 2
        ax.annotate(f"{count}", (bar.get_width(), y_pos), xytext=(2, 0), textcoords="offset points", va="center", fontsize=fontsize)
        ax.annotate(f" ({100 * count / total:.0f}%)", (bar.get_width(), y_pos), xytext=(2 + len(str(count)) * 6, 0), textcoords="offset points", va="center", fontsize=fontsize, color="gray")

ax.set_yticks(y)
ax.set_yticklabels(labels, fontsize=fontsize)
ax.set_xlabel("Count", fontsize=fontsize)
ax.set_xlim(0, 1100)
ax.tick_params(axis="x", labelsize=fontsize)
ax.legend(fontsize=fontsize, frameon=False)
ax.grid(False)

plt.tight_layout()
plt.savefig("./class_count_2_splits.svg", bbox_inches="tight")
plt.show()
```

```python
palette = PALETTES["house_scape"]
split_colors = {"train": palette[0], "val": palette[2], "test": palette[1]}
fontsize = 10

def get_off_size(path):
    with open(path, "r") as f:
        first = f.readline().strip()
        counts = f.readline().strip().split() if first == "OFF" else first[3:].strip().split()
    return int(counts[0]), int(counts[1])

def build_df(data_path, splits):
    rows = []
    for class_dir in sorted(data_path.iterdir()):
        if not class_dir.is_dir():
            continue
        for split in splits:
            split_dir = class_dir / split
            if not split_dir.exists():
                continue
            for path in split_dir.glob("*.off"):
                vertices, faces = get_off_size(path)
                rows.append({"Class": class_dir.name.replace("_", " ").title(), "Split": split.title(), "Vertices": vertices, "Faces": faces})
    return pd.DataFrame(rows)

def plot_mesh_stats(df, splits, save_path):
    split_order = [s.title() for s in splits]
    colors = {s.title(): split_colors[s] for s in splits}

    fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

    sns.boxplot(data=df, x="Class", y="Vertices", hue="Split", hue_order=split_order, palette=colors, ax=axes[0])
    sns.boxplot(data=df, x="Class", y="Faces", hue="Split", hue_order=split_order, palette=colors, ax=axes[1])

    for ax in axes:
        ax.set_yscale("log")
        ax.yaxis.set_major_locator(ticker.LogLocator(base=10))
        ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: rf"$10^{{{int(np.log10(x))}}}$"))
        ax.yaxis.set_minor_locator(ticker.LogLocator(base=10, subs=(2, 5)))
        ax.yaxis.set_minor_formatter(ticker.NullFormatter())
        ax.grid(False)
        ax.set_xlabel("")
        ax.tick_params(axis="both", which="major", labelsize=fontsize)
        ax.set_ylabel(ax.get_ylabel(), fontsize=fontsize)

    axes[0].legend(title=None, frameon=False, fontsize=fontsize)
    axes[1].legend_.remove()

    plt.xticks(rotation=0, fontsize=fontsize)
    plt.tight_layout()
    plt.savefig(save_path, bbox_inches="tight")
    plt.show()

df_2 = build_df(DATA_PATH, ["train", "test"])
plot_mesh_stats(df_2, ["train", "test"], "./mesh_statistics_2_splits.svg")

df_3 = build_df(DATA_PATH_3_splits, ["train", "val", "test"])
plot_mesh_stats(df_3, ["train", "val", "test"], "./mesh_statistics_3_splits.svg")
```

**Two-split:** *Distribution of mesh resolution across ModelNet10 classes in the training and test sets, quantified by the number of vertices and faces per mesh. Values are shown on a logarithmic scale.*

**Three-split:** *Distribution of mesh resolution across ModelNet10 classes in the training, validation, and test sets, quantified by the number of vertices and faces per mesh. Values are shown on a logarithmic scale.*


A logarithmic y-axis is useful when the mesh sizes vary by several orders of magnitude. Here, equal vertical distances represent **multiplication**, not addition:

* (10^2 = 100)
* (10^3 = 1{,}000)
* (10^4 = 10{,}000)
* (10^5 = 100{,}000)

So moving from (10^3) to (10^4) means **10× more vertices/faces**. Likewise, `500 → 5,000 → 50,000` would appear roughly equally spaced because each step is also a factor of 10.

For reading your boxplots, a median near (10^3) means about **1,000**, while one near (10^4) means about **10,000**. The smaller ticks help identify intermediate values such as (2\times10^3=2{,}000) and (5\times10^3=5{,}000).

For readers unfamiliar with logarithmic scales, Khan Academy has a very accessible explanation and video. ([Khan Academy][1])

```python
categories_files = defaultdict(list)
for d in DATA_PATH.iterdir():
    if not d.is_dir():
        continue

    categories_files[d.name] = np.array(list((d / "train").glob("*.off")))
```

```python
# Select examples
seed = 11

np.random.seed(seed)

files_to_visualise = defaultdict(list)
for cat in categories_files.keys():
    files_to_visualise [cat] = np.random.choice(categories_files[cat], 3)
```

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots


def visualize_rotate_grid(
    data_dict,
    sample_names=None,
    num_frames=100,
    frame_duration=50,
    figsize=None,
    aspectmode="cube",
):
    """
    Grid of rotating 3-D point clouds using pcshow-style markers.

    Parameters
    ----------
    data_dict : dict[str, list[np.ndarray]]
        Categories as keys, lists of (N, 3) arrays as values.
        Each column = one key. Each row = one sample in the list.
    sample_names : dict[str, list[str]], optional
        Subplot titles. Keys must match data_dict.
    num_frames : int
        Frames per full rotation.
    frame_duration : int
        Milliseconds per frame.
    figsize : tuple[int, int] | None
        (width, height) in pixels. Auto-calculated if None.
    aspectmode : str
        Plotly scene aspect mode ("cube", "data", etc.).
    """
    if not data_dict:
        raise ValueError("data_dict cannot be empty")

    categories = list(data_dict.keys())
    n_cols = len(categories)
    n_rows = max(len(v) for v in data_dict.values())

    # Default sample names
    if sample_names is None:
        sample_names = {
            cat: [f"Sample {i + 1}" for i in range(len(data_dict[cat]))]
            for cat in categories
        }

    # Build subplot titles in row-major order
    subplot_titles = []
    for r in range(n_rows):
        for c in range(n_cols):
            cat = categories[c]
            names = sample_names.get(cat, [])
            subplot_titles.append(names[r] if r < len(names) else "")

    fig = make_subplots(
        rows=n_rows,
        cols=n_cols,
        specs=[[{"type": "scene"}] * n_cols for _ in range(n_rows)],
        subplot_titles=subplot_titles,
        column_titles=categories,
        horizontal_spacing=0.02,
        vertical_spacing=0.12,
    )

    camera_radius = 1.25
    camera_z = 0.8

    # ------------------------------------------------------------------
    # Add traces with pcshow-style markers
    # ------------------------------------------------------------------
    for c_idx, cat in enumerate(categories):
        for r_idx in range(n_rows):
            scene_num = r_idx * n_cols + c_idx + 1
            scene_name = "scene" if scene_num == 1 else f"scene{scene_num}"

            if r_idx < len(data_dict[cat]):
                pts = np.asarray(data_dict[cat][r_idx])

                if pts.ndim != 2 or pts.shape[1] != 3:
                    raise ValueError(
                        f"data_dict['{cat}'][{r_idx}] shape {pts.shape}, expected (N, 3)"
                    )

                # pcshow-style trace
                trace = go.Scatter3d(
                    x=pts[:, 0],
                    y=pts[:, 1],
                    z=pts[:, 2],
                    mode="markers",
                    marker=dict(
                        size=2,
                        line=dict(
                            width=2,
                            color="DarkSlateGrey",
                        ),
                    ),
                )
                fig.add_trace(trace, row=r_idx + 1, col=c_idx + 1)

            fig.layout[scene_name].update(
                camera=dict(eye=dict(x=camera_radius, y=0, z=camera_z)),
                aspectmode=aspectmode,
            )

    # ------------------------------------------------------------------
    # Build rotation frames
    # ------------------------------------------------------------------
    angles = np.linspace(0, 2 * np.pi, num_frames, endpoint=False)
    scene_names = [
        "scene" if i == 0 else f"scene{i + 1}"
        for i in range(n_rows * n_cols)
    ]

    frames = []
    for theta in angles:
        layout_patch = {}
        for sname in scene_names:
            layout_patch[sname] = {
                "camera": {
                    "eye": {
                        "x": camera_radius * np.cos(theta),
                        "y": camera_radius * np.sin(theta),
                        "z": camera_z,
                    }
                }
            }
        frames.append(go.Frame(layout=layout_patch))

    fig.frames = frames

    # ------------------------------------------------------------------
    # Layout
    # ------------------------------------------------------------------
    if figsize is None:
        figsize = (max(280 * n_cols, 600), max(280 * n_rows, 400))

    fig.update_layout(
        width=figsize[0],
        height=figsize[1],
        showlegend=False,
        updatemenus=[
            {
                "type": "buttons",
                "showactive": False,
                "direction": "left",
                "pad": {"r": 10, "t": 10},
                "x": 0.1,
                "y": -0.05,
                "buttons": [
                    {
                        "label": "▶ Play",
                        "method": "animate",
                        "args": [
                            None,
                            {
                                "frame": {"duration": frame_duration, "redraw": True},
                                "transition": {"duration": 0},
                                "fromcurrent": True,
                                "mode": "immediate",
                            },
                        ],
                    },
                    {
                        "label": "⏸ Pause",
                        "method": "animate",
                        "args": [
                            [None],
                            {
                                "frame": {"duration": 0, "redraw": False},
                                "mode": "immediate",
                                "transition": {"duration": 0},
                            },
                        ],
                    },
                ],
            }
        ],
    )

    return fig
```

```python
point_cloud_data = defaultdict(list)
faces_data = defaultdict(list)
for cat, paths in files_to_visualise.items():
    for p in paths:
        mesh = trimesh.load(p)
        point_cloud = mesh.sample(4096)
        faces = mesh.faces
        point_cloud_data[cat].append(point_cloud)
        faces_data[cat].append(faces)
```
